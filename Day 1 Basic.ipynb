```python
#  Arithmetic operators
a = 5 + 5
b = 5 - 5
c = 5 / 5
d = 5 // 5
e = 5 * 5
f = 5 ** 5
g = 5 % 5

print("addition : ",a)
print("subtraction : ",b)
print("division : ",c)
print("integer part : ",d)
print("multiplication : ",e)
print("Power : ",f)
print("Remainder : ",g)
```

```python
# Camparision operator 
print(5 == 5)   # Equal 
print(5!= 2)    # Not Equal
print(5 < 1)     # less than  
print(5 > 3)    # Greater than 
print(3 <= 5)     # less than or equal
print(10 >= 5)    # Greter than or equal
```

```python
# logical operator
sal = 500000
location ="Noida"

# logical and 
if (sal >= 500000 and location =="Noida"):
    print("Yes, I will accept")
else: 
    print( "Not accepted")
```

```python
# logical operator
sal = 500000
location ="Noida"

# logical OR
if (sal > 500000 or location =="Delhi"):
    print("Yes, I will accept")
else: 
    print( "Not accepted")
```
