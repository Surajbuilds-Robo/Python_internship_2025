```python
# step 1: Accept the marks of student
marks = []
for i in range (5):
    mark = int(input(f"Enter marks of student {i +1} "))
    marks.append(mark)
marks

pass_count =0
fail_count =0

if mark >+40:
    pass_count +=1
else:
    fail_count += 1
print(f"Passed student {pass_count} and Failed student {fail_count}")

topers_marks = max(marks)
lowest_marks = min(marks)
print("Toppers marks :",topers_marks)
print("lowest marks :",lowest_marks)
```

```python
import random

# Step 1: List of animals
animals = ['elephant', 'tiger', 'giraffe', 'zebra', 'kangaroo', 'monkey', 'panda', 'lion', 'rabbit', 'hippopotamus']

# Step 2: Randomly select an animal
secret_animal = random.choice(animals)

# Step 3: Show a hint
print("Welcome to the Animal Guessing Game!")
print("Hint: The animal's name starts with '{}' and has {} letters.".format(secret_animal[0], len(secret_animal)))

# Step 4: Limited attempts
attempts = 3

# Step 5: Loop with condition
while attempts > 0:
    guess = input("Enter your guess: ").lower()
    
    if guess == secret_animal:
        print(" Correct! You guessed the animal.")
        break
    else:
        attempts -= 1
        print("Incorrect guess. Attempts left:", attempts)

# Step 6: Final message
if attempts == 0:
    print("Game Over! The correct animal was:", secret_animal)
```
