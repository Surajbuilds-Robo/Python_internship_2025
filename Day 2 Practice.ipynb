```python
P = int(input("Principal loan amount : "))
R = float(input("Annual interest rate (in %) : "))
T = int(input("Loan tenure (in months) : "))

r = R / (12 * 100)


EMI = (P * r * (1 + r) ** T) / ((1 + r) ** T - 1)

print("Monthly EMI to be paid : ₹", EMI)
```
