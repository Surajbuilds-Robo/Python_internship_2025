```python
for i in range(0,4):
    for j in range(1,8):
        if(j>=4-i and j<=4+i):
            print("*", end=' ')
        else:
            print(" ",end=" ")
    print("")
```

```python
rows = int(input("Enter the number of rows: "))
for i in range(1, rows + 1):
    print(" " * (rows - i) + "* " * i)

for j in range(rows - 1, 0, -1):
    print(" " * (rows - j) + "* " * i)
```
