```python
resume_text = """
I am a passionate software developer with experience in Python, Java, and SQL. 
I have worked on several machine learning projects and am familiar with Git,sql ,Git bash
Excellent communication skills and a basic understanding of AWS.
"""

required_skills = ["python", "java", "machine learning", "sql", "git", "communication", "aws"]

resume_text = resume_text.lower()

matched_skills = []
missing_skills = []

for skill in required_skills:
    if skill in resume_text:
        matched_skills.append(skill)
    else:
        missing_skills.append(skill)

print("Matched skills:", matched_skills)
print("Missed skills:", missing_skills)

match_percentage = len(matched_skills) / len(required_skills) * 100
print(f"\nMatched Percentage: {match_percentage:.2f}%")
```
