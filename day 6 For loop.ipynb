```python
#number guessing game
import random
print("welcome in number guessing game")
secret_num=random.randint(1,100)
print("computer is thinking a number between 1 to 100")
attempt=int(input("enter number of attempts you want to play "))
for i in range(1,attempt+1):
    guess_num=int(input(f"attempt{i}:guess a number"))
    if(guess_num<secret_num):
        print("too low")
    elif(guess_num>secret_num):
        print("too high")
    else:
        print(f"congrats you win in {i} attempt")
        break
print(f"your attempt is over !the secret number was {secret_num} play again")
```

```python
#accept no. from user and calculate
#1.total no. of digits
#2.sum of digits
#3.reverse a number 
#4.check no. is palidrome
#1.
n=int(input("enter number "))
count=0
while(n>0):
    count=count+1
    n=n//10
print(count)
```

```python
#3. reverse
n=int(input("enter the number"))
rev=0
while n>0:
    rev=rev*10 + n%10
    n//=10
print ("reversed number:",rev)
```

```python
#2. sum 
n=int(input("enter the number"))
sum=0
while n>0:
    r=n%10
    sum=sum+r
    n//=10
print("sum of digit is:" ,sum)
```

```python
#4. pallindrome
n=int(input("enter the number"))
rev=0
while n>0:
    rev=rev*10 + n%10
    n//=10
print ("reversed number:",rev)
if (rev==n):
    print("its pellindrome number")
else:
    print("not pellindrome number")

```
