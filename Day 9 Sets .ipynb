```python
# We can have a list in a set
vowels = [ 'a' , 'e' , 'i' ,'o' ,  'o' , 'u' , 'x' ]
vowels_set = set(vowels)
vowels_set
```

```python
# Adding a single element
vowels_set.add('hello')
vowels_set
```

```python
# adding multiple elements
# ADDING A LIST
vowels_set.update(['hi' , 'welcome'])
vowels_set
```

```python
# ADDING A TUPLE
vowels_set.update( ('r' , 'm') )
vowels_set
```

```python
vowels_set.discard('p')
vowels_set
```

```python
# discard() --- if the item does not exist in the set it will remain UNCHANGED
# remove() ---- if the item does not exist in the set it will show ERROR

vowels_set.discard("m")
vowels_set
```

```python
vowels_set.remove('hello')
vowels_set
```

```python
# pop() --- removing any random element and will also show the element 
print(vowels_set.pop())
```

```python
# difference_update()
A = {'a','c','g','d'}
B = {'c','f','g'}
result =A.difference_update(B)

print("A =",A)
print("B =",B)
print("result",result)
```

```python
# intersection_update()
A = {1,2,3,4}
B = {2,3,4,5}
result =A.intersection_update(B)
print("A =",A)
print("B =",B)
print("result",result)
```
