## Data types in  python

```python
# integer
i = 25 
print("Value of i : ",i)
print("Type of i : ",type(i))
```

```python
# float
f = 25.2 
print("Value of i : ",f)
print("Type of i : ",type(f))
```

```python
# negitive float no
a= -25.2
print(a )
```

```python
# complex no
y = 56 + 6j
print("Value of Y : ",y)
print("Type of Y : ", type(y))
```

```python
# String 
name ="Suraj"
b ='''Hello'''
print("Name : ", name)
print("Msg : ", b)
print("Type & Length of : ",type(name), len(name)) 
print("Type of : ",type(b)) 
# Sliceing
print(name[0:2])
```

```python
# Boolan
a = True
print(a)
print(type(a))
```

```python
# list 
li = ["heman",5, 9 , "java"]
print( li)
print("Type & Length of : ",type(li), len(li)) 
li = ["hema",5, 9 , "ham"]
print( li)
print("Type & Length of : ",type(li), len(li)) 
```

```python
# Tuple
tup =(5,5,"KAme")
print((type(tup)), (tup))
```

```python
# dictionary
name_dept = {"rekha" : "B.tech CSE",
            "hemant": "B.tech"            
            } 
print(name_dept.keys())
print(name_dept,"\n",type(name_dept))
```

```python
g={9,5,6,8}
print(g)
print(type(g))
```
