```python
#list comprehension-- it is a concise way to create a list using single line code
#square of numbers from 1 to 5
[x**2 for x in range (1,6)]
```

```python
#even numbers from 1 to 20
even_numbers= [x for x in range(1,21) if x%2==0]
even_numbers
```

```python
#convert all items of list in upper case 
names=['manvi','anjali','yash','suraj']
upper_case=[x.upper() for x in names]
upper_case
       
```

```python
#create a list of length of words
names=['manvi','anjali','yash','suraj']
length=[len(x) for x in names]
length
```

```python
# extract digit from a given string
text="abc123def456"
extract_digit=[x for x in text if x.isdigit()] 
extract_digit
```

```python
# dictionary comprehension---creating dictionary in single line
#square of numbers
squares={x:x**2 for x in range(1,6)}
squares
```

```python
#even numbers
nums= { x : x**2 for x in range(1,21) if(x%2 == 0) }
nums
```

```python
#multiple of 3 between 1 to 50 
nums={x:x**2 for x in range(2,50) if x%3==0}
nums
```

```python
#create dictionary fron two list 
roll_numbers=[1056,1021,1164,1134]
names=['anjali','palak','yash','lavisha']
roll_names={roll:name for roll, name in zip(roll_numbers,names)}
roll_names
```

```python
#filter dictionary 
marks={'maths':41,'english':95,'science':85,'sst':50}
{sub:mark for sub, mark in marks.items() if mark>50}
```

```python
#filter the marks of science
{sub:mark for sub, mark in marks.items() if sub=='science'}
```

```python
# count occurances of characters in a word
name="geeta university"
count={char:name.count(char) for char in set(name)}
count
```

```python
#packing and unpacking 
#unpacking values in list
my_list=[10,20,30,40]
a,b,c,d=my_list
print("a =",a)
print("b =",b)
print("c =",c)
print("d =",d)
```

```python
# using * for extended unpacking in lists
a,*b=my_list
print("a=",a)
print("b=",b)
```

```python
# unpacking in a function arguments
def add(a,b,c):
    return a+b+c
my_list=[10,20,30]
print(add(*my_list))
```

```python
#list packing in a function parameters(using arguements)
def greet_all(*names):
    for name in names:
        print("hello",name)
greet_all("alice","bob","atul","anjali","palak","manvi")
```

```python
# packing and unpacking in dictionary
# unpacking a dictionary
my_dict={
     'name':'anjali',
     'age':20,
     'city':'delhi'}
print(my_dict)
```

```python
name,age,city=my_dict.values()
print(name,age,city)
```

```python
#using ** to unpack dictionaries into functions
def display(name, age, city):
    print(f"name:{name},age:{age},city:{city}")
data={'name':'alice','age':25,'city':'delhi'}
display(**data)
```

```python
# dictionary packing with **kwargs
def student_info(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
student_info(name="anjali",grade="a",subject="maths")
```

```python
# merging dictionaries with ** unpacking
dict1={'a':1, 'b':2}
dict2={'c':3, 'd':4}
merged={**dict1,**dict2}
print(merged)
```
