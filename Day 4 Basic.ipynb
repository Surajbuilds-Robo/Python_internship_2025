```python
#check whether a given character is vowel or consonent
ch= input("enter the character").lower()
if (ch=='a') or (ch=='e') or (ch=='i') or (ch=='o') or (ch=='u'):
    print(f"you entered {ch} is a vowel")
else:
    print(f"you entered {ch} is a consonent")
```

```python
#write a program to take two number and an operator from user to calculate the result
num1=int(input("enter a number"))
num2=int(input("enter another number"))
op = input ("enter the operator -->+,-,*,/,%")
if op=='+':
      print("num1 +num2=", num1 + num2)
elif op=='-':
      print("num1 - num2=", num1 - num2)
elif op=='*':
      print("num1 * num2=", num1 * num2)
elif op=='/':
      print("num1 / num2=", num1 / num2)
elif op=='%':
      print("num1 % num2=", num1 % num2)
else:
     print("invalid operation")
```
