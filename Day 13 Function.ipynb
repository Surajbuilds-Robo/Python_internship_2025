```python
# Arbitary argument : passing of mutltiple value  for single argument
def greet(*names):
    for i in names:
        print(f"Good morning {i}")

greet("Radha", "Gopal", "Charu","Seeta",4455)
```

```python
# Passing collection into  function:

def change_List(nums):
    val =10
    new_list =[]
    for i in range(0,len(nums)):
        new_list.append(nums[i]+val)
    print('After function call list :',new_list)


nums = [5,10,15,20]
change_List(nums)
```

```python
# Find the square of even number form 10 to 20 ussing passing collections
# as an argument in a function in python

def square(nums):
    square=[]
    for i in nums:
        if i%2==0:
            square.append(i**2)
    return square
    
num_list = list(range(1, 21))

result = square(num_list)


print("Squares of even numbers from 1 to 20:", result)
```

```python
# Passing function to function
def value(r):
    return r
def circle_area(f):
    area =3.14 *(f**2)
    return area
res = circle_area(value(4))
print('Area of circle : ',res)
```

## Anonymous function

```python
double = lambda x: x*2
print(double(5))
```

```python
square = lambda x: x**2
print(square(5))
```

```python
greater = lambda x,y :x  if x>y else y
greater(5,10)
```

```python
even = lambda x: "Even" if x%2==0 else "odd"
print(even(12))
```

```python
# Filter
my_list =[1,5,4,6,8,11,3,12]
new_list = list(filter(lambda x: (x%2==0),my_list))
print(new_list)
```

```python
# map()-------->
my_list =[1,5,4,6,8,11,3,12]
new_list = list(map(lambda x: x*2,my_list))
print(new_list)
```
